```python
import numpy as np
import qutip as qt
```

```python
def X_p2epsilon(epsilon):
    return (-1j*(np.pi/2+epsilon)/2*qt.sigmax()).expm()

def Z(delta):
    return (-1j*delta/2*qt.sigmaz()).expm()

def Y_p2epsilon(epsilon):
    return (-1j*(np.pi/2+epsilon)/2*qt.sigmay()).expm()
```

```python
epsilon = 0.2
delta = np.pi/2

Z(-delta)*X_p2epsilon(epsilon)*Z(delta)
```

```python
def new_unitary(epsilon, delta):
    return Z(-delta)*X_p2epsilon(epsilon)*Z(delta)
```

```python
new_unitary(epsilon, delta)
```
